# Task 1 - Data Cleaning and Preprocessing


## Load Dataset

```python
import pandas as pd

df = pd.read_csv("dataset/raw_data.csv", sep="\t")

df.head()
```

## Dataset Info

```python
# Dataset info
df.info()
```

## Missing Values

```python
df.isnull().sum()
```

## Duplicate Values

```python
df.duplicated().sum()
```

## Clean Column Names

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ", "_")
```

## Date Format Conversion

```python
df["dt_customer"] = pd.to_datetime(
    df["dt_customer"],
    dayfirst=True
)

df["dt_customer"].head()
```

## Verify Data Types

```python
df.dtypes
```

## Final Validation

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

## Save Final Cleaned Dataset

```python
df.to_csv(
    "dataset/cleaned_data.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
